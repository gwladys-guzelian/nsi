!!! info Objectifs

Dans ce TP, nous allons :

 - paramétrer la communication d'un ensemble d'ordinateurs au sein de réseaux locaux,
 - paramétrer la communication entre différents réseaux locaux,
 - installer un serveur Web et accéder à la page hébergée dessus,
 - installer un serveur DNS et configurer les ordinateurs pour l'interroger.

Nous allons utiliser pour cela un outil de simulation : [Filius](https://www.lernsoftware-filius.de/Herunterladen) est un logiciel libre permettant de reproduire le fonctionnement de réseaux.

!!!


!!! note À faire

Ouvrir le logiciel filius et enregistrer immédiatement le document ouvert dans le dossier SNT de votre répertoire personnel.


!!!

## 1. Créer un réseau local

!!! note À faire

- Ajouter **3** ordinateurs
- Donner à chacun une adresse IP unique
- Cocher la case "Utiliser l'adresse IP comme nom"

## 2. Connecter deux réseaux locaux

!!! note À faire

 - Ajouter un autre réseau local constitué de deux PC et d'un switch.
 - Configurer les adresses IP des PC.
 - Ajouter un routeur comportant **3 interfaces**.
 - Faire un ping vers une machine de ce nouveau réseau local (depuis une machine du premier réseau).

!!!

??? tip Vidéo : ajouter un routeur

!!! question Question 3
Pourquoi la commande `ping` vers une machine d'un autre réseau échoue-t-elle ?
!!!

!!! note À faire

 - Configurer les adresses IP du routeurs de façons à ce qu'elles correspondent à celles des différents réseaux.
 - Configurer les PC en renseignant l'adresse de leur passerelle.
 - Tester de nouveau la commande ping entre deux ordinateurs de différents réseaux.

## 3. Ajouter et interroger un serveur Web

!!! note À faire

 - Vérifier l'adresse de la passerelle soit correctement configurée pour **tous les PC**.
 - Lancer la simulation et ajouter un serveur Web sur un PC (n'oublier pas de le démarrer).
 - Ajouter un navigateur sur un PC de l'autre réseau.
 - Ouvrir le navigateur et entrer dans celui-ci l'adresse IP du serveur Web pour consulter la page hébergée.

!!!

## 4. Installer et configurer un serveur DNS

!!! note À faire
 - Ajouter un nouveau PC et le nommer "serveur DNS"
 - Relier ce PC au routeur
 - Installer sur ce PC un serveur DNS
 - Démarrer le serveur DNS
 - Ajouter un nouveau nom de domaine (par exemple *toto.com* ) et y associer l'adresse IP du serveur Web
 - Tester l'accès à *toto.com* depuis un des PC.
!!!
