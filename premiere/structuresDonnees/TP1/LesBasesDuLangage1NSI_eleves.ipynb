!!! quote
Python possède une fonction **id()** qui renvoie l’adresse mémoire de la variable donnée en argument. 
- Ecraser la valeur de a avec une nouvelle valeur et redemander l’adresse de a. 
- L’adresse de a n’est plus la même. Que s’est-il passé ? 

Le stockage des variables en mémoire ne fonctionne donc pas comme une place unique réservée pour stocker les valeurs successives d’une variable <br><br>

L’affectation `a = 12` ne provoque pas la réservation définitive d’un espace-mémoire pour la variable `a` mais la création d’un lien vers un espace-mémoire qui contient la valeur 12.<br><br>

Si on crée une nouvelle variable `b` aussi égale à 12, elle va aussi pointer sur la même adresse-mémoire.
!!!

````python
a = 32
print(id(a))
a = 12
print(id(a))
b = 12
print(id(b))
````

!!! quote
Les variables `a` et `b` renvoient vers la même adresse-mémoire.

Python pré-positionne les entiers de –5 à 256 sur des adresses-mémoires car il considère que ces entiers servent souvent et doivent donc être rapidement accessibles.

Affectons maintenant à `a` la valeur 2026. Que se passe-t-il ?

Si on créé une nouvelle variable `c` qui vaut aussi 2026, Python va ouvrir une autre adresse-mémoire pour stocker 2026 alors qu’il l’a déjà stocké ailleurs.

Attention, l’adresse mémoire de deux variables ayant la même valeur peut parfois être différente.

```python
a = 2026
print(id(a))
c = 2026
print(id(c))
```

### À retenir

Une variable peut changer de valeur au cours de l'exécution.

```python
score = 10
score = score + 5
```

La nouvelle valeur de `score` est `15`.
!!!

!!! example
Observons l’instruction
a = 2 + 3

- **Etape 1 : évaluation** <br>
Python va prendre la partie à droite du signe égal et va l'évaluer, ce qui signifie qu'il va essayer de lui donner une valeur. Dans nos exemples, cette valeur sera numérique, mais elle peut être d'un autre type.
Ici, Python effectue le calcul 2 + 3 et l'évalue à la valeur 5.<br><br>
- **Etape 2 : affectation** <br>
Une fois évaluée l'expression à droite du signe =, il ne reste plus qu'à l'affecter à la variable (déjà existante ou pas) située à gauche du signe =.
Comme expliqué précédemment, un «lien» est fait entre le nom de la variable et l'adresse-mémoire qui contient la valeur évaluée. a sera donc lié à la valeur 5. Plus simplement, on dira que «a vaut 5»

!!!

!!! example
**Incrémenter = augmenter**
Soit une variable appelée `compteur` initialisée à 0 (au démarrage du programme).
`compteur = 0` 

A un moment du programme, cette variable change en lui ajoutant 1. Cela donne en Python : 

`compteur = compteur + 1` 

On dit que compteur a été incrémenté de 1.

**Interprétation de l’instruction du point de vue de l’informaticien :**

- On évalue la partie droite de l’égalité 
- On va chercher le contenu de la variable compteur. Si celle-ci n’existe pas, un message d’erreur est renvoyé.
- On additionne 1 au contenu de la variable compteur.
- On écrase le contenu actuel de la variable compteur avec le contenu obtenu à l’étape précédente.

On peut aussi écrire :

**Compteur += 1**

# 12. Mettre au point une fonction avec un jeu de tests

Un **jeu de tests** est un ensemble de cas choisis pour vérifier qu'une fonction respecte sa spécification.

| Type de cas | Rôle |
|:---|:---|
| Cas nominal | Vérifier le fonctionnement habituel |
| Cas limite | Tester une valeur à la frontière |
| Cas particulier | Tester une situation spéciale |
| Cas d'erreur | Vérifier une donnée interdite si elle doit être gérée |


!!! example 
### Méthode

1. Lire la spécification.
2. Choisir des entrées pertinentes.
3. Prévoir les résultats attendus.
4. Exécuter les tests.
5. Comparer les résultats.
6. Corriger la fonction si nécessaire.
!!!

### Exemple — Utiliser `assert`

````python
def maximum(a, b):
    if a > b:
        return a
    return b

assert maximum(8, 3) == 8
assert maximum(2, 7) == 7
assert maximum(5, 5) == 5

print("Tous les tests sont réussis.")
````

!!! example
L'instruction `assert` vérifie qu'une condition est vraie.

```python
assert resultat_obtenu == resultat_attendu
```


Si elle est fausse, Python déclenche une `AssertionError`.
!!!

# 13. L'importation

Il existe trois façons d'utiliser un module en Python.

| Syntaxe | Effet |
|:--------|:------|
| `import math` | Importe le module `math`. Les fonctions s'utilisent avec `math.` (`math.sqrt(2)`). |
| `from math import sqrt` | Importe uniquement la fonction `sqrt`. On l'utilise directement (`sqrt(2)`). |
| `from math import *` | Importe toutes les fonctions et constantes du module. Cette syntaxe est déconseillée car elle peut créer des conflits de noms. |

👉 **Bonne pratique :** privilégier `import math` ou `from math import ...`.

````python
import math

print(math.sqrt(2))
print(math.pi)
````

````python
from math import sqrt

print(sqrt(2))
````

!!! example
### Importer un module

#### Importer le module

```python
import math

print(math.sqrt(2))
print(math.pi)
```

Le module `math` est importé. Les fonctions et constantes sont utilisées en les précédant de `math.`.

---

#### Importer seulement certaines fonctions

```python
from math import sqrt

print(sqrt(2))
```

Seule la fonction `sqrt` est importée.

---

#### Importer toutes les fonctions du module (à éviter)

```python
from math import *

print(sqrt(2))
print(pi)
```

Toutes les fonctions et constantes du module `math` deviennent directement accessibles.

En effet, cette syntaxe est généralement déconseillée car :
- elle importe tous les noms dans l'espace de noms courant ;
- on ne sait plus d'où proviennent les fonctions ;
- elle peut provoquer des conflits de noms.
