Pour créer la 1ère table, on écrit ceci en langage SQL (pensez à exécuter les cellules à chaque fois) :

````sql
/* création de la table livre*/
DROP TABLE IF EXISTS livre;            -- au cas où une table livre existerait déjà
CREATE TABLE livre (                   -- DROP TABLE livre supprime la table livre (là on le fait seulement si elle existe)
    id INTEGER PRIMARY KEY NOT NULL,       -- on définit l'attribut id comme clef primaire
    titre TEXT NOT NULL,                   -- le NOT NULL signifie qu'on n'accepte pas d'absence de valeur pour tel attribut
    id_auteur INTEGER REFERENCES auteur (id) NOT NULL,  -- on définit l'attribut id_auteur comme clef étrangère
    ann_publi INTEGER,
    note INTEGER,
    CHECK (note>=0 AND note<=10)           -- pour spécifier que la note devra être comprise entre 0 et 10
    );
````

### Remarque.
*Le langage SQL ne tient pas du tout compte de la casse (autrement dit, pour lui minuscule=minuscule). Pour plus de lisibilité, on mettra les mots clés en majuscule et le reste en minuscule.*

## Fonctions d'agrégation

Les fonctions d'agrégation permettent d'appliquer une fonction à l'ensemble des valeurs d'une colonne (ou de plusieurs colonnes) et de renvoyer le résultat comme une table à 1 seule ligne et 1 seule colonne.

Voici quelques fonctions d'agrégations :
+ `COUNT` permettant de compter le nombre d'entrées non vide
+ `AVG` (pour <em>average</em> en anglais qui signifie moyenne) permettant de calculer la moyenne d'une colonne
+ `SUM` permettant de calculer la somme d'une colonne
+ `MIN` et `MAX` permettant de trouver respectivement le minimum et le maximum d'une colonne 

Dans une telle requête, il faut bien décomposer les choses pour construire la requête : d'abord, le résultat qui va faire un certain nombre de lignes, puis la fonction d'agrégation qui est appliquée sur ce résultat. 

Exécutez la cellule suivante et vérifiez qu'elle affiche bien le nombre de lignes de la table livre

````sql
SELECT COUNT (*) FROM livre;
````

On préfère renommer le titre avec le mot clé `AS`

````sql
SELECT COUNT (*) AS Total FROM livre;
````

On peut vouloir obtenir la note moyenne des livres : 

````sql
SELECT AVG (note) AS note_moyenne FROM livre;
````

ou bien obtenir le nombre d'auteurs anglais dans la table auteur

````sql
SELECT COUNT (*) AS nombre_d_auteurs_anglais FROM auteur WHERE langue_ecrit='anglais';
````

ou bien connaître le nombre de langues distinctes de la table auteur

````sql
SELECT COUNT (DISTINCT langue_ecrit) AS nombre_de_langues FROM auteur;
````

## Jointures

Il arrive que l'information sur une entité soit dispersée dans deux tables. On doit alors utiliser une jointure. 

Par ex. quels sont les titres des livres écrits par K. Dick ?

Quand l'information est dispersée dans deux tables T1 et T2, la jointure consiste à constuire d'abord une sorte de grosse table où chaque ligne est composée d'une ligne de T1 et d'une ligne de T2 (on appelle cette étape le produit cartésien de T1 avec T2). Cette grosse table contient le nombre de lignes dans T1 multiplié par le nombre de lignes dans T2. 

````sql
SELECT * FROM livre JOIN auteur;
````

Le produit cartésien contient encore beaucoup trop d'informations et une telle jointure n'a pas grand intérêt. 

En face de chaque ligne d'un livre, on veut seulement l'auteur correspondant (et non pas comme précédemment tous les auteurs). 

En effet, les colonnes *id_auteur* de la table *livre* et *id* de la table *auteur* ont les mêmes valeurs. On peut alors faire une jointure sur les valeurs des ces colonnes similaires, qui vont servir de point de branchement. 

````sql
SELECT * FROM livre JOIN auteur ON livre.id_auteur=auteur.id; -- on précise sur quoi on fait la jointure --
````

C'est déjà beaucoup plus satisfaisant. Mais on peut vouloir ne pas afficher les colonnes avec les identifiants

````sql
SELECT livre.titre,livre.ann_publi,livre.note,auteur.nom,auteur.prenom,auteur.ann_naiss,auteur.langue_ecrit
FROM livre
JOIN auteur ON livre.id_auteur=auteur.id;
````

On peut aussi mettre des alias avec le mot clé `AS` pour racourcir l'instruction

````sql
SELECT l.titre,l.ann_publi,l.note,a.nom,a.prenom,a.ann_naiss,a.langue_ecrit
FROM livre AS l
JOIN auteur AS a ON l.id_auteur=a.id;
````

On peut vouloir afficher tous les livres écrits par K.Dick

````sql
SELECT l.titre,l.ann_publi,l.note
FROM livre AS l
JOIN auteur AS a ON l.id_auteur=a.id
WHERE a.nom='K. Dick';
````

**Remarque.** Il est possible de faire des jointures sur plus de 2 tables, il suffit de rajouter des `JOIN <table> ON <condition de jointure>`.

On peut vouloir calculer la moyenne des notes des livres de K. Dick

````sql
SELECT AVG (l.note) AS note_moyenne
FROM livre AS l
JOIN auteur AS a ON l.id_auteur=a.id
WHERE a.nom='K. Dick';
````

## 4. Comment modifier des données ?

Les données stockées dans une SGBD ne sont *a priori* pas figées et peuvent être modifiées au cours du temps. Nous allons montrer deux types de modifications pouvant être faites sur les tables :
- la suppression d'un ensemble de lignes
- la mise à jour de certains attributs d'un ensemble de lignes

### La suppression des lignes avec `DELETE FROM <table> WHERE <condition>`
Permet de supprimer toutes les lignes de la table vérifiant la condition donnée.

````sql
DELETE FROM livre WHERE ann_publi>1950;
````

Vérifions que l'instruction précédente a bien supprimé de la table livre toutes les lignes dont l'année de publication est strictement supérieure à 1950 :

````sql
SELECT * FROM livre;
````

### Attention à ne pas confondre les instructions `DELETE FROM <table>` et `DROP TABLE <table>`
- La 1ère supprime toutes les lignes de la table mais la table existe et on peut donc insérer des éléments,
- La 2nde supprime totalement la table.

### La mise à jour avec `UPDATE <table> SET <attribut>=<nouvelle_valeur> WHERE <condition>`
Sélectionne toutes les lignes de la table vérifiant la condition donnée et, pour chacune de ces lignes, remplace la valeur courante de l'attribut donné par la nouvelle valeur.

````sql
UPDATE auteur SET langue_ecrit='english' WHERE langue_ecrit='anglais';
````

Vérifiez que la table *auteur* a été correctement mise à jour en exécutant la cellule suivante

````sql
SELECT * FROM auteur;
````
