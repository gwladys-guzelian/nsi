## Les commandes utiles

### Ajout d'une nouvelle entité
```sql
INSERT INTO
nom_de_la_table 
VALUES 
(valeur1,valeur2,...);        

````sql
INSERT INTO Lyceen VALUES (672,"Marrani","Yael","13/03/2005","Sec");
````

<div style="color:#000000;background-color:#ddffaa;padding:10px; border: solid #88bb55; border-width:1px 1px 1px 5px; border-radius:5px 5px;">

On peut ajouter plusieurs entitée à l'aide d'une seule commande en séparant les tuples représentant les entités par des virgules.

````sql
INSERT INTO Lyceen VALUES
 (673,"Elbaz","Léa","08/09/2005","Sec"),
 (674,"Koher","Sonia","20/07/2005","Sec");
````

<div style="color:#000000;background-color:#ddffaa;padding:10px; border: solid #88bb55; border-width:1px 1px 1px 5px; border-radius:5px 5px;">

Avec cette syntaxe, les valeurs doivent être fournies dans l'ordre de déclaration des attributs dans la relation.           
On peut également préciser l'ordre dans lequel les valeurs sont fournies, en indiquant les noms des attributs correspondant, dans le même ordre : 

````sql
INSERT INTO Lyceen (niveau, id,  ddn, prenom, nom) 
VALUES  ("Sec", 675, "22/11/2005","Lubin","Harmont");
````

<div style="color:#000000;background-color:#ddffaa;padding:10px; border: solid #88bb55; border-width:1px 1px 1px 5px; border-radius:5px 5px;">

### Modification des valeurs d'attribut pour une ou plusieurs entités existantes

```sql
UPDATE
nom_de_la_table
SET attribut=nouvelle_valeur
WHERE condition;

````sql
UPDATE Lyceen SET nom="Merrani" WHERE id=672;
````

````sql
UPDATE Lyceen SET id=id+100 WHERE niveau="Sec";
````

<div style="color:#000000;background-color:#ddffaa;padding:10px; border: solid #88bb55; border-width:1px 1px 1px 5px; border-radius:5px 5px;">

### Suppression d'une ou plusieurs entités
```sql
DELETE
FROM nom_de_la_table
WHERE condition;

````sql
DELETE FROM Lyceen WHERE id=674;
````

### Maintien de la cohérence de la base de données.
**Les modifications ne peuvent être exécutées que si elles ne contreviennent pas aux contraintes d'intégrité**. 
Une modification sur une table contenant une clé étrangère peut nécessiter des interventions sur les tables qui font référence à cette clé, afin de respecter la contrainte de référence.
