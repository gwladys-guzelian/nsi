<h2>Quelques requêtes sur plusieurs tables</h2>

Avant de pouvoir effectuer des requêtes sur plusieurs tables nous devons prendre connaissance de la structure globale de notre base de données.



Les champs soulignés sont des _clés primaires_, et les champs précédés d'un # sont des _clés étrangères_.

!!! info Définition :

La clé étrangère représente un champ (ou des champs) qui pointe vers la clé primaire d’une autre table. 
!!!
Le modèle relationnela avec les liens :




Prenons un exemple :

Je souhaite connaitre le nom de l'univers auquel appartient Black Widow. Pour obtenir cette information nous devrions commencer par trouver l'```idunivers``` du personnage dans la table ```personnage```

!!! example  Example

```sql
SELECT idunivers FROM personnage
WHERE nom = 'Black Widow';
```

!!!


Nous trouvons : 1.

Ensuite, une seconde requête dans la table ```univers``` devrait nous permettre de répondre à la question :

!!! example  Example
```sql
SELECT nom FROM univers
WHERE idunivers = '1';
```
!!!
Et nous trouvons que Black Widow appartient à l'univers Marvel.

Il est possible de combiner ces deux requêtes en une seule, c'est ce qu'on appelle des requêtes imbriquées.

!!! example  Example

```sql
SELECT nom FROM univers
WHERE idunivers = (SELECT idunivers FROM personnage
                    WHERE nom = 'Black Widow');
```
!!!

!!! example  Exemple

Donnons les personnages mauvais

```sql
SELECT personnage.nom FROM personnage
JOIN alignement ON alignement.idalignement = personnage.idalignement
WHERE alignement.nom = 'Mauvais';
    
```
